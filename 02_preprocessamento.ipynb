## Tratamento das variáveis representadas em dias

As colunas `DAYS_BIRTH` e `DAYS_EMPLOYED` estavam representadas em dias. Para facilitar a interpretação, `DAYS_BIRTH` foi convertida para idade aproximada em anos, criando a variável `AGE_YEARS`.

A coluna `DAYS_EMPLOYED` foi analisada separadamente, pois contém o valor especial `365243`, que não representa um tempo real de emprego. Esse valor foi sinalizado por meio da variável `EMPLOYED_SPECIAL_VALUE`, substituído por `NA` em `DAYS_EMPLOYED_TREATED` e utilizado para criar a variável `EMPLOYED_YEARS`.

```python
import pandas as pd
import numpy as np

# 1. Criar idade em anos
app["AGE_YEARS"] = (-app["DAYS_BIRTH"] / 365).round(1)

# 2. Contar quantas vezes aparece o valor especial 365243 em DAYS_EMPLOYED
qtd_special_value = (app["DAYS_EMPLOYED"] == 365243).sum()
print("Quantidade de registros com DAYS_EMPLOYED = 365243:", qtd_special_value)

# 3. Criar flag para o valor especial de DAYS_EMPLOYED
app["EMPLOYED_SPECIAL_VALUE"] = (app["DAYS_EMPLOYED"] == 365243).astype(int)

# 4. Criar versão tratada de DAYS_EMPLOYED
app["DAYS_EMPLOYED_TREATED"] = app["DAYS_EMPLOYED"].replace(365243, pd.NA)

# 5. Criar tempo de emprego em anos
app["EMPLOYED_YEARS"] = (
    -pd.to_numeric(app["DAYS_EMPLOYED_TREATED"], errors="coerce") / 365
).round(1)

# 6. Converter Y/N para 1/0
# np.where evita o FutureWarning do replace e mantém o código simples.
# A condição considera "Y", 1 e "1" como sim, o que torna a célula segura
# caso ela seja executada mais de uma vez.
app["FLAG_OWN_CAR"] = np.where(
    app["FLAG_OWN_CAR"].isin(["Y", 1, "1"]),
    1,
    0
)

app["FLAG_OWN_REALTY"] = np.where(
    app["FLAG_OWN_REALTY"].isin(["Y", 1, "1"]),
    1,
    0
)

# 7. Conferir conversão das variáveis binárias
print("\nDistribuição de FLAG_OWN_CAR:")
print(app["FLAG_OWN_CAR"].value_counts(dropna=False))

print("\nDistribuição de FLAG_OWN_REALTY:")
print(app["FLAG_OWN_REALTY"].value_counts(dropna=False))
```

## Criação da variável auxiliar `BAD_MONTH`

A partir da coluna `STATUS`, foi criada a variável auxiliar `BAD_MONTH`, indicando se determinado registro mensal representa um atraso grave.

Foram considerados como atraso grave os status `2`, `3`, `4` e `5`, que correspondem a atrasos iguais ou superiores a 60 dias. Dessa forma, registros com esses status receberam valor `1`, enquanto os demais receberam valor `0`.

## Criação da variável alvo `TARGET_BAD`

A variável alvo foi criada a partir da coluna auxiliar `BAD_MONTH`. Como `BAD_MONTH` indica se determinado mês teve atraso grave, foi necessário consolidar essa informação no nível do cliente.

Para isso, os registros foram agrupados por `ID`, utilizando o valor máximo de `BAD_MONTH`. Dessa forma, se o cliente apresentou pelo menos um mês com atraso grave, recebeu `TARGET_BAD = 1`; caso contrário, recebeu `TARGET_BAD = 0`.

Após a criação da variável alvo, foi verificada sua distribuição absoluta e percentual. Observou-se forte desbalanceamento, com proporção muito maior de clientes classificados como bons pagadores do que maus pagadores.

```python
# Criar conjuntos de IDs para comparar as bases
ids_app = set(app["ID"])
ids_target = set(target["ID"])

# Verificar interseção entre app e target
print("Clientes em app:", len(ids_app))
print("Clientes em target:", len(ids_target))
print("Clientes em ambas:", len(ids_app.intersection(ids_target)))
print("Clientes em app sem target:", len(ids_app - ids_target))
print("Clientes em target sem app:", len(ids_target - ids_app))
```

## Junção das bases `application_record` e `target`

Antes de unir as bases, foi verificada a interseção entre os IDs da base cadastral (`app`) e os IDs da variável alvo (`target`).

Observou-se que nem todos os clientes da base cadastral possuem histórico de crédito, e nem todos os clientes com histórico de crédito possuem cadastro disponível na base `application_record`. Por esse motivo, foi utilizado um `merge` do tipo `inner`, mantendo apenas os clientes presentes nas duas bases.

## Preparação final para modelagem

Após a análise exploratória dos dados, foram realizadas etapas adicionais de preparação da base para a fase de modelagem.

A primeira etapa consistiu na criação da variável `LOG_INCOME`, obtida por meio da transformação logarítmica da renda anual (`AMT_INCOME_TOTAL`). Durante a análise exploratória foi observado que a distribuição da renda apresenta forte assimetria à direita, com poucos clientes possuindo rendas muito superiores ao restante da população. A transformação logarítmica reduz o impacto desses valores extremos sem removê-los da base, preservando a informação e tornando a distribuição mais adequada para determinados algoritmos de classificação.

Em seguida, foi realizado o tratamento final da variável `EMPLOYED_YEARS`. Durante o pré-processamento inicial, os registros associados ao valor especial `365243` em `DAYS_EMPLOYED` foram convertidos para valores ausentes e sinalizados por meio da variável `EMPLOYED_SPECIAL_VALUE`.

Como esses registros representam uma condição específica já identificada e documentada, os valores ausentes de `EMPLOYED_YEARS` foram substituídos por `-1`. Esse valor foi escolhido por não representar um tempo de emprego real, permitindo diferenciar esses casos dos demais registros sem perder informação.

Após essas transformações, foi realizada uma nova verificação de valores ausentes para garantir que todas as variáveis utilizadas na etapa de modelagem estivessem completas.

```python
# Transformação logarítmica da renda
df["LOG_INCOME"] = np.log1p(df["AMT_INCOME_TOTAL"])

# Tratamento do valor especial de tempo de emprego
df["EMPLOYED_YEARS"] = df["EMPLOYED_YEARS"].fillna(-1)

# Conferência final dos valores ausentes
print(df.isna().sum().sort_values(ascending=False))
```
